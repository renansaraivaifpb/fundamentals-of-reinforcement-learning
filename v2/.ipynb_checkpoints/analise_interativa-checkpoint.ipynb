# 🔬 Análise Interativa de Parâmetros do Ambiente de Simulação

Este notebook é uma ferramenta para explorar interativamente o impacto dos parâmetros do ambiente e da função de recompensa no comportamento do nosso sistema de ar-condicionado.

```python
# Célula 1: Instalação e Importação de Bibliotecas
import ipywidgets as widgets
from ipywidgets import interactive
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd
import json
import os
from IPython.display import display, clear_output

# Importa as classes dos nossos outros arquivos
try:
    from classroom_ac_env import ClassroomACEnvironment, ClassroomConfig, ACState
    from ac_qlearning_agent import ACQLearningAgent
    from ac_dqn_agent import ACDQNAgent, DQNConfig
    # Reutilizamos funções do analyzer, então ele precisa estar na pasta
    from analyzer import load_agent_from_results, run_simulation, plot_scenario_dashboard 
    print("Módulos do projeto carregados com sucesso!")
except ImportError as e:
    print(f"ERRO ao importar módulos: {e}\nCertifique-se de que este notebook está na mesma pasta dos seus scripts .py")

sns.set_theme(style="whitegrid")
```

## Parte 1: Explorando a Física do Ambiente

Nesta seção, vamos simular a resposta da temperatura da sala a diferentes condições, **sem um agente de RL**. O objetivo é ganhar intuição sobre como os parâmetros físicos afetam o ambiente.

**Experimente:**
- **Aumente a `thermal_mass` (Inércia Térmica):** Veja como a temperatura muda de forma muito mais lenta.
- **Aumente o `heat_transfer_coeff` (Isolamento Ruim):** Veja como a temperatura interna tende a seguir a externa mais rapidamente.
- **Aumente a `occupancy` (Ocupação):** Observe o aquecimento gerado pelas pessoas.

```python
# Célula 2: Função para simular a física
def plot_environment_physics(thermal_mass, heat_transfer_coeff, occupancy, ac_action_str, season):
    config = ClassroomConfig(
        thermal_mass=thermal_mass,
        heat_transfer_coeff=heat_transfer_coeff,
        season=season
    )
    env = ClassroomACEnvironment(config)
    
    # Define as condições iniciais
    options = {'start_temp': 28.0, 'occupancy': occupancy, 'hour_of_day': 14}
    env.reset(options=options)
    
    ac_action = ACState[ac_action_str].value
    temps = [env.current_temp]
    outdoor_temps = [env._calculate_outdoor_temp()]
    
    for _ in range(120): # Simula 12 horas
        _, _, _, _, info = env.step(ac_action)
        temps.append(info['temperature'])
        outdoor_temps.append(env._calculate_outdoor_temp())
    
    plt.figure(figsize=(14, 7))
    plt.plot(temps, label='Temperatura Interna', lw=3, color='royalblue')
    plt.plot(outdoor_temps, label='Temperatura Externa', lw=2, color='gray', linestyle='--')
    plt.axhspan(config.temp_comfort_min, config.temp_comfort_max, color='green', alpha=0.15, label='Faixa de Conforto')
    plt.title(f'Resposta da Sala com AC em "{ac_action_str}" e {occupancy} Pessoas', fontsize=16)
    plt.ylabel('Temperatura (°C)')
    plt.xlabel('Passos de Tempo (6 min)')
    plt.legend()
    plt.ylim(15, 40)
    plt.grid(True, which='both', linestyle='--')
    plt.show()

# Cria os widgets interativos
interactive_plot = interactive(plot_environment_physics, 
                               thermal_mass=widgets.FloatSlider(value=1000, min=500, max=5000, step=100, description='Inércia Térmica:'),
                               heat_transfer_coeff=widgets.FloatSlider(value=0.5, min=0.1, max=2.0, step=0.1, description='Coef. de Calor:'),
                               occupancy=widgets.IntSlider(value=30, min=0, max=50, step=1, description='Ocupação:'),
                               ac_action_str=widgets.Dropdown(options=[s.name for s in ACState], value='MEDIUM', description='Ação do AC:'),
                               season=widgets.Dropdown(options=['summer', 'winter'], value='summer', description='Estação:')
                              )

display(interactive_plot)
```

## Parte 2: Explorando a Função de Recompensa

A função de recompensa é o que **motiva** o agente. Entender como diferentes componentes (bônus por conforto, penalidade por energia) afetam a recompensa final é crucial. 

**Experimente:**
- Aumente o `comfort_bonus`: veja como a recompensa por estar confortável se torna muito mais atrativa.
- Aumente o `energy_penalty_mult`: observe como a penalidade por usar o AC se torna mais severa, especialmente em horários de pico.

```python
# Célula 3: Função para visualizar a recompensa
def analyze_reward(comfort_bonus, warm_penalty, energy_penalty_mult, is_peak_hour):
    reward_structure = {'COMFORTABLE': comfort_bonus, 'WARM': warm_penalty}
    config = ClassroomConfig(reward_structure=reward_structure, energy_penalty_multiplier_peak=energy_penalty_mult)
    env = ClassroomACEnvironment(config)
    
    scenarios = {
        'Confortável (AC OFF)': {'temp': 24.0, 'ac_state': ACState.OFF},
        'Quente (AC MEDIUM)': {'temp': 28.0, 'ac_state': ACState.MEDIUM},
        'Quente (AC HIGH)': {'temp': 28.0, 'ac_state': ACState.HIGH}
    }
    
    print("--- Análise de Recompensa Instantânea ---")
    for name, params in scenarios.items():
        env.current_temp = params['temp']
        env.ac_state = params['ac_state']
        env.hour_of_day = 19 if is_peak_hour else 14 # 19h é pico, 14h não é
        
        # Calcula os componentes da recompensa
        comfort_level = env._get_comfort_level()
        comfort_rw = env.config.reward_structure.get(comfort_level.name, 0)
        
        energy_consumption = env.config.ac_energy_consumption[env.ac_state]
        multiplier = env.config.energy_penalty_multiplier_peak if is_peak_hour else 1.0
        energy_pen = -energy_consumption * 0.1 * multiplier
        
        total_reward = comfort_rw + energy_pen
        
        print(f"\nCenário: {name}")
        print(f"  Bônus/Punição de Conforto: {comfort_rw:.2f}")
        print(f"  Penalidade de Energia: {energy_pen:.2f}")
        print(f"  RECOMPENSA TOTAL: {total_reward:.2f}")

interactive_reward_analysis = interactive(analyze_reward,
                                          comfort_bonus=widgets.FloatSlider(value=1.0, min=1.0, max=20.0, step=1.0, description='Bônus Conforto:'),
                                          warm_penalty=widgets.FloatSlider(value=-5.0, min=-20.0, max=-1.0, step=1.0, description='Punição Calor:'),
                                          energy_penalty_mult=widgets.FloatSlider(value=1.5, min=1.0, max=5.0, step=0.5, description='Mult. Pico Energia:'),
                                          is_peak_hour=widgets.Checkbox(value=False, description='É Horário de Pico?')
                                         )
display(interactive_reward_analysis)
```

## Parte 3: Análise de Agentes Treinados

Finalmente, vamos carregar nossos agentes treinados e observar seu comportamento em diferentes cenários. Isso nos permite validar as políticas que eles aprenderam.

**Como Usar:**
1. Insira o caminho para a sua pasta de resultados (ex: `results_20251007_103409`).
2. Clique no botão **"Carregar Modelos"**.
3. Selecione o agente e o cenário que deseja analisar nos menus dropdown.
4. O gráfico será gerado automaticamente.

```python
# Célula 4: Widgets e lógica para análise de agentes

# Carrega os cenários do arquivo externo
try:
    with open('scenarios.json', 'r') as f:
        scenarios_list = json.load(f)
    scenario_options = {s['name']: s for s in scenarios_list}
except FileNotFoundError:
    print("Arquivo 'scenarios.json' não encontrado! Usando cenários de fallback.")
    scenario_options = {'fallback': {'name': 'fallback', 'start_temp': 28.0, 'occupancy': 30, 'hour': 14}}
    
# Widgets de interface
results_dir_input = widgets.Text(value='results_...', description='Pasta de Resultados:', layout={'width': '400px'})
load_button = widgets.Button(description="Carregar Modelos")
model_dropdown = widgets.Dropdown(description='Agente:')
scenario_dropdown = widgets.Dropdown(options=scenario_options.keys(), description='Cenário:')
output_area = widgets.Output()

# Dicionário global para guardar agentes carregados
loaded_agents = {}

# Função para carregar os modelos quando o botão é clicado
def on_load_button_clicked(b):
    with output_area:
        clear_output(wait=True)
        results_dir = results_dir_input.value
        if not os.path.isdir(results_dir):
            print(f"Diretório não encontrado: {results_dir}")
            return
        
        print(f"Procurando modelos em '{results_dir}'...")
        json_files = sorted(glob.glob(os.path.join(results_dir, '*_results.json')))
        model_names = [os.path.basename(f).replace('_results.json', '') for f in json_files]
        
        if not model_names:
            print("Nenhum arquivo de resultado encontrado.")
            return
            
        loaded_agents.clear()
        for name in model_names:
            # O nome do modelo no dropdown será mais legível
            display_name = name.replace('_', ' ').title()
            result = load_agent_from_results(results_dir, name)
            if result:
                loaded_agents[display_name] = result
        
        model_dropdown.options = loaded_agents.keys()
        print(f"Modelos carregados com sucesso: {list(loaded_agents.keys())}")

# Função para atualizar o gráfico quando um dropdown muda
def on_value_change(change):
    with output_area:
        # Só roda se os valores estiverem selecionados
        if not change['new'] or not model_dropdown.value or not scenario_dropdown.value:
            return
            
        clear_output(wait=True)
        print("Gerando simulação...")
        
        model_display_name = model_dropdown.value
        agent, env_config, _ = loaded_agents[model_display_name]
        
        scenario_name = scenario_dropdown.value
        scenario_params = scenario_options[scenario_name]
        
        env = ClassroomACEnvironment(env_config)
        df_result = run_simulation(agent, env, scenario_params)
        
        # Reutiliza a função de plotagem do analyzer
        plot_scenario_dashboard(model_display_name, {scenario_name: df_result}, env, ".")
        print("Simulação concluída.")
        
# Conecta as funções aos widgets
load_button.on_click(on_load_button_clicked)
model_dropdown.observe(on_value_change, names='value')
scenario_dropdown.observe(on_value_change, names='value')

# Exibe a interface
display(widgets.VBox([widgets.HBox([results_dir_input, load_button]), 
                       widgets.HBox([model_dropdown, scenario_dropdown]), 
                       output_area]))
```
